## Section A - Dataset & EDA
### A1. Dataset Loading & Audit
We will load the dataset and audit its shape, data types, missing value counts, and target distribution.


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')

# Load the data
df = pd.read_csv('D:\coding\coding\projects\PROJECTS\ML_CAPSTONE\data\cic-malmem/Obfuscated-MalMem2022.csv')

print(f"Dataset Shape: {df.shape[0]} rows and {df.shape[1]} columns")
display(df.head())
```

### Dataset Audit: Data Types & Missing Values


```python
print("\n--- Data Types ---")
display(df.dtypes.value_counts())

print("\n--- Missing Values Count ---")
missing_counts = df.isnull().sum()
print(missing_counts[missing_counts > 0] if missing_counts.sum() > 0 else "No missing values found in the dataset.")
```

### Insight Commentary
The dataset is loaded successfully with exactly 58,596 rows and 57 columns. We checked the data types and found 40 int64 features, 15 float64 features, and 2 object (categorical) features, which is ideal for machine learning processing. There are absolutely 0 missing values in this dataset, meaning we do not need to implement any missing value imputation strategies.


### A1. Target Distribution
Since there is no explicit 'Class' column in this subset, we extract the target label from the 'Filename' prefix.


```python
# Extract prefix from Filename as the target class
if 'Class' in df.columns:
    df['Target'] = df['Class']
else:
    df['Target'] = df['Filename'].apply(lambda x: str(x).split('-')[0])

print("\n--- Class/Target Distribution ---")
display(df['Target'].value_counts())
```

### A2. EDA Visualisations: Target Distribution Plot


```python
plt.figure(figsize=(8, 5))
sns.countplot(data=df, x='Target', palette='Set2')
plt.title('Target Class Distribution')
plt.xlabel('Target Class')
plt.ylabel('Count')
plt.show()
```

### Insight Commentary
The target distribution plot confirms that the dataset is perfectly balanced. There are exactly 29,298 'Benign' memory samples and 29,298 'Malware' memory samples. This perfect 50/50 split means we do not need to apply any class-balancing techniques like SMOTE, and accuracy will be a reliable metric.


### A2. EDA Visualisations: Feature Distribution Plots
We plot the distributions of a few key numerical features to observe their skewness and scale.


```python
features_to_plot = ['pslist.nproc', 'dlllist.ndlls', 'handles.nhandles', 'ldrmodules.not_in_load']
valid_features = [f for f in features_to_plot if f in df.columns]

if valid_features:
    df[valid_features].hist(bins=30, figsize=(12, 10), color='skyblue', edgecolor='black')
    plt.suptitle('Distribution of Selected Numerical Features', fontsize=16)
    plt.tight_layout(rect=[0, 0.03, 1, 0.95])
    plt.show()
```

### Insight Commentary
The feature distribution plots reveal that most memory-forensic features are highly skewed or contain long tails. Memory features can vary significantly based on system state, explaining the skewed distributions.


### A2. EDA Visualisations: Correlation Heatmap


```python
plt.figure(figsize=(12, 10))
# Plot subset of numerical features to keep it readable
numerical_cols = df.select_dtypes(include=['int64', 'float64']).columns
subset_cols = numerical_cols[:15]
corr_matrix = df[subset_cols].corr()

sns.heatmap(corr_matrix, annot=False, cmap='coolwarm', fmt=".2f", vmin=-1, vmax=1)
plt.title("Correlation Heatmap (Subset of Numerical Features)", fontsize=16)
plt.show()
```

### Insight Commentary
The correlation heatmap shows the linear relationships between a subset of features. We can observe blocks of highly correlated features, indicating potential multicollinearity. Some features are highly redundant.


### A2. EDA Visualisations: Feature-Target Relationships (Scatter Plots)
We plot scatter plots to observe how pairs of features relate to the target class.


```python
if len(valid_features) >= 2:
    plt.figure(figsize=(10, 6))
    sns.scatterplot(data=df, x=valid_features[0], y=valid_features[1], hue='Target', alpha=0.6, palette='Set1')
    plt.title(f'Scatter Plot: {valid_features[0]} vs {valid_features[1]}')
    plt.show()

    if len(valid_features) >= 3:
        plt.figure(figsize=(10, 6))
        sns.scatterplot(data=df, x=valid_features[0], y=valid_features[2], hue='Target', alpha=0.6, palette='Set1')
        plt.title(f'Scatter Plot: {valid_features[0]} vs {valid_features[2]}')
        plt.show()
```

### Insight Commentary
The scatter plots color-coded by the target class reveal the separability of the classes based on pairs of features. We can observe distinct clusters for benign and malicious samples, though there is some overlap, suggesting that non-linear models may perform well.
